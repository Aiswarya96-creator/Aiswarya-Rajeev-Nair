```python
cat_cols=shop.select_dtypes(include=['object']).columns
num_cols=shop.select_dtypes(include=['int64','float64']).columns

for col in num_cols:
  if shop[col].isnull().sum() > 0:
    shop[col]=shop[col].fillna(shop[col].median())

for col in cat_cols:
  if shop[col].isnull().sum() > 0:
    shop[col]=shop[col].fillna(shop[col].mode()[0])
print(cat_cols)
print(num_cols)
```

```python
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder

# 1. Define Ordinal Categories (features with a natural hierarchy or order)
frequency_order = ["Rarely", "Occasionally", "Often", "Always"]
list_order = ["Never", "Sometimes", "Often", "Always"]
spend_order = [
    "Under $50",
    "$50-150",
    "$150-300",
    "$300-500",
    "$500+",
]
income_order = ["Under $1,000", "$1,000-3,000", "$3,000-6,000", "$6,000+"]

# Encode Ordinal Columns
ordinal_cols = [
    "Online_Shopping_Frequency",
    "Makes_Shopping_List",
    "Avg_Monthly_Spend_NonEssentials",
    "Monthly_Income_Range",
]

ord_enc = OrdinalEncoder(
    categories=[frequency_order, list_order, spend_order, income_order],
    handle_unknown="use_encoded_value",
    unknown_value=-1,
)

shop[ordinal_cols] = ord_enc.fit_transform(shop[ordinal_cols])

# 2. Identify Nominal Categories (features without intrinsic order for One-Hot Encoding)
nominal_cols = [
    "Gender",
    "Occupation_Status",
    "Country_Region",
    "Shop_Most_Where",
    "Preferred_Platform",
    "Primary_Device",
    "Top_Spending_Category",
    "Return_Frequency",
    "Preferred_Payment_Method",
    "Reason_Prefer_Online",
    "Reason_Prefer_InStore",
]

# One-Hot Encode nominal features
shop = pd.get_dummies(shop, columns=nominal_cols, drop_first=True)

# 3. Verify encoding (All columns should now be int, float, or bool)
print("Remaining object columns:", shop.select_dtypes(include=["object"]).columns.tolist())
```

```python

if "Timestamp" in shop.columns:
    shop = shop.drop(columns=["Timestamp"])

# 2. Separate features (X) and target variable (y)
# Drop both the unencoded target text column and the previously encoded version
X = shop.drop(columns=["Best_Shopping_Mode", "Best_Shopping_Mode_encoded"], errors="ignore")
y = shop["Best_Shopping_Mode_encoded"]

# 3. Verify all feature columns in X are numeric
print("Data types in X:")
print(X.dtypes.value_counts())
print("\nAny non-numeric remaining in X?:", X.select_dtypes(include=["object"]).columns.tolist())
```

```python


from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import classification_report, accuracy_score, precision_recall_fscore_support, confusion_matrix, ConfusionMatrixDisplay

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

# ==========================================
# 1. PREPROCESSING & DATA PREPARATION
# ==========================================
# Convert boolean columns to integer 1/0
bool_cols = shop.select_dtypes(include=['bool']).columns
shop[bool_cols] = shop[bool_cols].astype('int64')

# Label encode target variable
le = LabelEncoder()
shop['Best_Shopping_Mode_encoded'] = le.fit_transform(shop['Best_Shopping_Mode'])

# Define feature matrix X and target y
X = shop.drop(columns=['Best_Shopping_Mode', 'Best_Shopping_Mode_encoded', 'Timestamp'], errors='ignore')
y = shop['Best_Shopping_Mode_encoded']

# Split data (80% train, 20% test)
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)

# Scale numeric columns for distance/linear models
num_scale_cols = [
    'Age', 'Review_Influence_Score', 'Social_Ads_Influence_Score',
    'Price_Comparison_Frequency', 'Discount_Importance_Score',
    'Likely_To_Buy_After_Ad_Score', 'Impulse_Purchase_Frequency',
    'Online_Shopping_Satisfaction_Score', 'Trust_In_Online_Reviews_Score',
    'Regret_After_Purchase_Frequency'
]

scaler = StandardScaler()
x_train_scaled = x_train.copy()
x_test_scaled = x_test.copy()

x_train_scaled[num_scale_cols] = scaler.fit_transform(x_train[num_scale_cols])
x_test_scaled[num_scale_cols] = scaler.transform(x_test[num_scale_cols])

# ==========================================
# 2. MODEL TRAINING & EVALUATION
# ==========================================
models = {
    "Logistic Regression": (LogisticRegression(max_iter=1000, random_state=42), True),
    "Decision Tree": (DecisionTreeClassifier(random_state=42), False),
    "Random Forest": (RandomForestClassifier(n_estimators=100, random_state=42), False),
    "Gradient Boosting": (GradientBoostingClassifier(random_state=42), False)
}

results = []
model_objects = {}
predictions = {}

for name, (model, requires_scaling) in models.items():
    # Use scaled data for Logistic Regression, raw split for trees
    X_tr = x_train_scaled if requires_scaling else x_train
    X_te = x_test_scaled if requires_scaling else x_test

    model.fit(X_tr, y_train)
    y_pred = model.predict(X_te)

    acc = accuracy_score(y_test, y_pred)
    prec, rec, f1, _ = precision_recall_fscore_support(y_test, y_pred, average='weighted')

    results.append({
        "Model": name,
        "Accuracy": acc,
        "Precision": prec,
        "Recall": rec,
        "F1-Score": f1
    })

    model_objects[name] = model
    predictions[name] = y_pred

# Display comparison table
metrics_df = pd.DataFrame(results).sort_values(by="F1-Score", ascending=False)
print("--- MODEL PERFORMANCE COMPARISON ---")
display(metrics_df.round(4))

# ==========================================
# 3. CONFUSION MATRIX FOR BEST MODEL
# ==========================================
best_model_name = metrics_df.iloc[0]["Model"]
best_predictions = predictions[best_model_name]

print(f"\nDisplaying Confusion Matrix for the Best Model: {best_model_name}")

cm = confusion_matrix(y_test, best_predictions)
fig, ax = plt.subplots(figsize=(6, 5))
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=le.classes_)
disp.plot(cmap='Blues', ax=ax, values_format='d')
plt.title(f"Confusion Matrix ({best_model_name})", fontweight='bold')
plt.grid(False)
plt.show()

# ==========================================
# 4. TOP FEATURE IMPORTANCES
# ==========================================
# Extract importances from tree-based model (Random Forest / Gradient Boosting)
best_tree_model = model_objects.get("Random Forest", model_objects.get("Gradient Boosting"))

importances = best_tree_model.feature_importances_
feature_imp = pd.DataFrame({'Feature': X.columns, 'Importance': importances})
feature_imp = feature_imp.sort_values(by='Importance', ascending=False).head(10)

plt.figure(figsize=(9, 5))
sns.barplot(data=feature_imp, x='Importance', y='Feature', palette='viridis')
plt.title('Top 10 Features Influencing Best Shopping Mode', fontsize=12, fontweight='bold')
plt.xlabel('Feature Importance Score')
plt.ylabel('Feature Name')
plt.tight_layout()
plt.show()
```
